```python
import os
import requests
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\allis\Downloads\Student Retention Data.csv")
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(data=df, x="Returned the following semester", y="GPA")
plt.title("")
plt.xlabel("Returned")
plt.ylabel("GPA")
plt.show()
```
